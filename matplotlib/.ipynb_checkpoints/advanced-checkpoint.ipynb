```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

### coloured scatterplots

```python
iris = pd.read_csv('iris.csv')
iris.head(30)
```

```python
iris['Species'] = iris['Species'].replace({'Iris-setosa':0, 'Iris-versicolor':1,'Iris-virginica':2})
iris.sample(5)
```

```python
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species']) # this 'c' attribute will differentiate species based on colors
plt.xlabel('sepal length')
plt.ylabel('petal length')
plt.colorbar()
```

### plot size

```python
plt.figure(figsize=(15,5))  # used to define size of graph

plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species']) # this 'c' attribute will differentiate species based on colors
plt.xlabel('sepal length')
plt.ylabel('petal length')
plt.colorbar()
```

### Annotations


```python
batters = pd.read_csv('batter.csv')
batters
```

```python
sample_df = batters.head(100).sample(25,random_state=5)
# here from top 100 batters we will select 25 random batters
sample_df
```

```python
plt.figure(figsize=(18,10))

plt.scatter(sample_df['avg'],sample_df['strike_rate'], s=sample_df['runs'])  # this s will show which batter has made how many run by dot size
# here we don't know which dot represents which batter

for i in range(sample_df.shape[0]):
    plt.text(sample_df['avg'].values[i],sample_df['strike_rate'].values[i],sample_df['batter'].values[i])

# this loop will help in identifying batters
    
```

### horizontal and vertical lines

```python
plt.figure(figsize=(18,10))

plt.scatter(sample_df['avg'],sample_df['strike_rate'], s=sample_df['runs']) 

plt.axhline(130,color='red')
plt.axvline(30,color='red')

for i in range(sample_df.shape[0]):
    plt.text(sample_df['avg'].values[i],sample_df['strike_rate'].values[i],sample_df['batter'].values[i])
```

### subplots

```python
# a diff way to plot graphs

batters.head()
```

```python
plt.figure(figsize=(10,3))
plt.scatter(batters['avg'],batters['strike_rate'])
plt.title("something")
plt.xlabel("avg")
plt.ylabel("strike_rate")
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10,3))

ax.scatter(batters['avg'],batters['strike_rate'])
ax.set_title("something")
ax.set_xlabel("avg")
ax.set_ylabel("strike_rate")
fig.show()

# we can use this way to plot multiple graphs side by side which is not possible in normal syntax
```

```python
fig,ax = plt.subplots(nrows=2,ncols=1,sharex=True) # sharex will make x-axis common for both graphs

ax[0].scatter(batters['avg'],batters['strike_rate'],color='red')
ax[1].scatter(batters['avg'],batters['runs']) 

ax[0].set_title("avg vs strike rate")
ax[1].set_title("avg vs runs")

ax[0].set_ylabel("strike rate")
ax[1].set_ylabel("runs")

ax[1].set_xlabel("avg")
```
