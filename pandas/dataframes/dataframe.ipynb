```python
# examples of drop duplicates

# find the last match played by virat kohli in ipl in delhi

ipl
```

```python
# first we will create a row of all players in a match

ipl['all_players'] = ipl['Team1Players'] + ipl['Team2Players']
ipl.head()
```

```python
# find virat in those all_players value

def did_kohli_play(player_list):
    return 'V Kohli' in player_list

ipl['all_players'].apply(did_kohli_play)
```

```python
# matches played by kohli in delhi

ipl['did_kohli_play'] = ipl['all_players'].apply(did_kohli_play)

mask = (ipl['City'] == 'Delhi') & (ipl['did_kohli_play'])

ipl[mask]
```

```python
# only the most recent match

ipl[mask].drop_duplicates(subset=['City','did_kohli_play'], keep='first')
```

```python
# apply() in dataframes

points_df = pd.DataFrame({
    '1st point': [(3, 4), (-6, 5), (0, 0), (-10, 1), (4, 5)],
    '2nd point': [(-3, 4), (0, 0), (2, 2), (10, 10), (1, 1)]
})

points_df
```

```python
def euclidean(row):
    pt_A = row['1st point']
    pt_B = row['2nd point']

    return ((pt_A[0] - pt_B[1])**2 + (pt_A[1]-pt_B[1])**2)**0.5

points_df['distance'] = points_df.apply(euclidean,axis=1)

points_df
```
