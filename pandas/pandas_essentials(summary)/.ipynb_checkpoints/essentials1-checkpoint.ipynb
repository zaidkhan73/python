### <ins>create a dataframe</ins>


```python
data = {
    "Name": ["Jhon","Arthur","richie"],
    "Age": [31,25,32],
    "City": ["Denmark","New york","Helsinkie"]
}

df = pd.DataFrame(data)
print(df)

# save the dataframe into file
df.to_csv("custom_data.csv", index=False)  # we wrote index=False to remove the numbering from the data
#similary to save in excel format: df.to_xlsx("custom_data.xlsx")
#similarly to save in json format: df.to_json("custom_data.json")
```

### <ins>handling of data in terms of rows</ins>

#### head()



this method is used to show rows of the beginning only

the head() can also accept a parametre if needed to show specific number of rows
eg: print(df.head(2)) this statement will give only 2 rows in output

```python
df = pd.read_excel("SampleSuperstore.xlsx")
print("first row")
print(df.head(1)) # if not given any parametre than 5 rows will be printed
```

#### tail()

the tail method is used to show the number of rows from the bottom similar to head() just opposite ends

```python
print("last row")
print(df.tail(1))  #here the last row is printed
```

### <ins>descriptive statistics (summary) of data frame</ins>

#### it tells you about:
- count (if there are missing values than count will be less than of rows)
- mean value of columns
- standard deviation of coulmns (if deviation high -> there is high deviation (difference) in data values in a column
- 25%, 50%, 75% of data in columns (calculated after sorting data in ascending order)
- minimum value of columns
- max value of columns|

```python
data = {
    "Name": [
        "John", "Arthur", "Richie", "Michael", "David", "Chris",
        "Robert", "Daniel", "James", "Andrew", "Kevin", "Thomas"
    ],
    "Age": [
        31, 25, 32, 28, 35, 29,
        40, 26, 34, 30, 27, 38
    ],
    "Salary": [
        55000, 48000, 62000, 50000, 75000, 53000,
        90000, 46000, 70000, 58000, 49000, 82000
    ],
    "Performance_Score": [
        8.2, 7.5, 8.8, 7.9, 9.1, 8.0,
        9.3, 7.2, 8.6, 8.1, 7.6, 9.0
    ]
}

df = pd.DataFrame(data)
print("sample data: ")
print(df,"\n")


print("descriptive statistics: ")
print(df.describe())
```

### <ins>filtering in dataframes</ins>

operations to do:
- select specific column
- filter rows
- combine multiple conditions
  

#### column selection:

```python
single_col = df["Name"]
multiple_col = df[["Name","Age","Salary"]]
print("single column selection: ")
print(single_col,"\n")
print("multiple coumn selection: ")
print(multiple_col)
```

#### row selection

```python
#normal selection
single_row = df["Age"]
multiple_row = df[["Name","Age"]]
print("single row selection: ")
print(single_row,"\n")
print("multiple row selection: ")
print(multiple_row,"\n")
# based on single condition:
filtered_rows = df[df["Salary"] > 50000]
filtered_rows2 = df[(df["Salary"] > 50000) & (df["Age"] < 30)] # here we can also use OR (|) condition
print("selections based on single and multiple conditions: ")
print(filtered_rows)
print(filtered_rows2)
```
