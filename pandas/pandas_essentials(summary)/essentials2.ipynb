# <ins>Modifying Data in Pandas Effectively</ins>

#### learning about:
- how to add data
- how to update data
- how to remove data

### <ins>adding columns</ins>

```python
data = {
    "Name": [
        "John", "Arthur", "Richie", "Michael", "David", "Chris",
        "Robert", "Daniel", "James", "Andrew", "Kevin", "Thomas"
    ],
    "Age": [
        31, 25, 32, 28, 35, 29,
        40, 26, 34, 30, 27, 38
    ],
    "Salary": [
        55000, 48000, 62000, 50000, 75000, 53000,
        90000, 46000, 70000, 58000, 49000, 82000
    ],
    "Performance_Score": [
        8.2, 7.5, 8.8, 7.9, 9.1, 8.0,
        9.3, 7.2, 8.6, 8.1, 7.6, 9.0
    ]
}

df = pd.DataFrame(data)
print(df,"\n \n")

# # using bracket
# df["bonus"] = df["Salary"] * 0.1   # 10% of salary
# print("added new column of bonus")
# print(df)

# using insert() method (recommended):
# df.insert(index,column_name,data_value)
df.insert(0,"employee_id",[f"EMP{str(i).zfill(3)}" for i in range(1, len(df) + 1)])
print("adding employee_id column")
print(df)
```

### <ins>updating data</ins>

```python
print(df,"\n \n")

#using loc[]
#df.loc[row_index,column_name] = new_value
# df.loc[0,'Salary',] = 80000
# print("updating salary of EMP")
# print(df)

# updating multiple values in data
df['Salary'] = df['Salary'] * 1.05  # here we are updating data of a column not addding a new column
print("increasing salary of employees: ")
print(df)
```

### <ins>Removing data</ins>

```python
print(df,"\n \n")

# using df.drop(Columns = ["column_name"], inplace = True)
# inplace = True -> modifying the original dataframe
# inplace = Fase -> making a copy of dataFrame and making changes in that

df.drop(columns=["Performance_Score"],inplace= True)
print(df)
```

# <ins>Handling missing data in Pandas</ins>

#### represented in two ways:

- NaN (not a number)
- None (for object data types)


### <ins>detecting missing data</ins>


```python
# using isNull()
# True -> value is missing
# False -> value is present

data = {
    "Name": [
        "John", None, "Richie", "Michael", "David", "Chris",
        "Robert", "Daniel", "James", "Andrew", "Kevin", "Thomas"
    ],
    "Age": [
        31, None, 32, 28, 35, 29,
        40, 26, 34, 30, 27, 38
    ],
    "Salary": [
        55000, None, 62000, 50000, 75000, 53000,
        90000, 46000, 70000, 58000, 49000, 82000
    ],
    "Performance_Score": [
        8.2, None, 8.8, 7.9, 9.1, 8.0,
        9.3, 7.2, 8.6, 8.1, 7.6, 9.0
    ]
}

df = pd.DataFrame(data)
print(df, "\n \n")

# see if data is missing
print("check for missing values: ")
print(df.isnull())  #in whichever field true is present there the data is missing in the data frame
print("\n \n number of missing values:\n ",df.isnull().sum())
```

### removing missing data cells

```python
df = pd.DataFrame(data)
print(df, "\n \n")


#removing missing data cells (dropna())
df.dropna(inplace=True)  # here we can give 'axis' to remove missing values specifically from rows or columns
print("after removing missing values: ")
print(df)
```

### filling missing data cells

```python
print(df)

# filling missing data cells(fillna())
df.fillna(0,inplace = True)
print("\n \n filling the missing values: \n")
print(df)
```

### <ins>interpolation in pandas</ins>

interpolation is the technique to fill missing values in dataframes in a smart and convinient way

---

examples:
10 - 20 - NaN - 40 - 50 (here by interpolation we can estimate that the values should be filled with 300)

---

why to use:

- preserve data integrity
- smooth trends
- avoid data loss

---

commonly used types:

- linear
- polynomial
- time

### linear interpolation

```python
data = {
    "Time": [1,2,3,4,5],
    "Value": [10,None,30,None,50]
}

df = pd.DataFrame(data)
print("before linear interpolation: \n")
print(df,"\n \n")

df["Value"] = df["Value"].interpolate(method="linear")
print("After linear interpolation: \n")
print(df)
```

# <ins>Sorting and Aggregation in Pandas</ins>

### <ins>sorting data</ins>

```python
data = {
    "Name": [
        "John", "Arthur", "Richie", "Michael", "David", "Chris",
        "Robert", "Daniel", "James", "Andrew", "Kevin", "Thomas"
    ],
    "Age": [
        31, 25, 32, 28, 35, 29,
        40, 25, 34, 30, 27, 38
    ],
    "Salary": [
        55000, 480000, 62000, 50000, 75000, 53000,
        90000, 46000, 70000, 58000, 49000, 82000
    ],
    "Performance_Score": [
        8.2, 7.5, 8.8, 7.9, 9.1, 8.0,
        9.3, 7.2, 8.6, 8.1, 7.6, 9.0
    ]
}

df = pd.DataFrame(data)
print(df,"\n \n")

# sorting data by a single column

df.sort_values(by="Age",ascending = False,inplace=True)  # False -> sort in descending order / True -> sort in ascending order
print("sorted age by descending order: \n ")
print(df,"\n \n")

# sorting by multiple columns

df.sort_values(by=["Age","Salary"], ascending = [True,True], inplace = True)
# here sorting is done on basis of age, if age is same than on basis of salary
print("sorted data by age and salary in ascending order: \n")
print(df)
```

### <ins>grouping of columns</ins>

#### groupby -> aggregate (sum / mean / max / min / count / std)

```python
data = {
    "Name": [
        "John", "Arthur", "Richie", "Michael", "David", "Chris",
        "Robert", "Daniel", "James", "Andrew", "Kevin", "Thomas"
    ],
    "Age": [
        31, 25, 28, 28, 30, 29,
        45, 25, 34, 30, 27, 30
    ],
    "Salary": [
        55000, 480000, 62000, 50000, 75000, 53000,
        90000, 46000, 70000, 58000, 49000, 82000
    ],
    "Performance_Score": [
        8.2, 7.5, 8.8, 7.9, 9.1, 8.0,
        9.3, 7.2, 8.6, 8.1, 7.6, 9.0
    ]
}

df = pd.DataFrame(data)
print(df,"\n \n")

grouped_salary_sum = df.groupby("Age")["Salary"].sum()

print(grouped_salary_sum)
```

# <ins>Merging and Joining</ins>


### merging rows of two data frames with same column

```python
# syntax ->
# pd.merge(df1, df2, on="column_name", how="type of join")

# customer dataframe
df_customers = pd.DataFrame({
    'Customer_Id': [1,2,3],
    'Name': ['John','Arthur','David']
})
print("customers dataframe: \n")
print(df_customers,"\n \n")

# order dataframe
df_orders = pd.DataFrame({
    'Customer_Id':[1,2,5],
    'Order_Amount':[250,450,350]
})
print("orders dataframe: \n")
print(df_orders,"\n \n")

# merge
df_merged = pd.merge(df_customers, df_orders, on="Customer_Id", how="inner")
print("merged dataframes (inner join): \n")
print(df_merged,"\n \n")

df_merged = pd.merge(df_customers, df_orders, on="Customer_Id", how="outer")
print("merged dataframes (outer join): \n")
print(df_merged,"\n \n")

df_merged = pd.merge(df_customers, df_orders, on="Customer_Id", how="left")
print("merged dataframes (left join): \n")
print(df_merged,"\n \n")

df_merged = pd.merge(df_customers, df_orders, on="Customer_Id", how="right")
print("merged dataframes (right join): \n")
print(df_merged,"\n \n")
```

### concatenation of dataframes 

joining data of two dataframes vertically(row-wise) or horizontally(column-wise)

```python
# syntax ->
# pd.concate([df1, df2], axis=0, ignore_index=True)

df_region1 = pd.DataFrame({
    'Customer_ID':[1,2,3],
    'Name':['John','Arthur','David']
})

df_region2 = pd.DataFrame({
    'Customer_ID':[4,5,6],
    'Name':['Michael','Richy','James']
})

# joining vertically:
df_concat = pd.concat([df_region1, df_region2], ignore_index=True) # by default axis = 0
print("joining vertically: \n")
print(df_concat, "\n \n")

#joining horizontally:
df_concat = pd.concat([df_region1, df_region2], axis=1)
print(df_concat)
```
